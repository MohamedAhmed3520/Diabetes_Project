```python
import pandas as pd
import numpy as np
```

```python
diab = pd.read_csv(r'C:\Users\Hamada\Downloads\archive (16)\Dataset of Diabetes .csv')
diab.head(5)
```

```python
diab.drop('ID', axis=1, inplace=True)
```

```python
diab.info()
```

```python
diab.duplicated().sum()
```

```python
diab['CLASS'].value_counts()
```

```python
diab['Gender'] = diab['Gender'].map({'F': 0, 'M': 1})
diab['CLASS'] = diab['CLASS'].map({'N': 0, 'Y': 1 , 'P': 2 })
```

```python
diab.head(5)
```

```python
diab['CLASS'].value_counts()
```

```python
diab['CLASS'].isnull().sum()
```

```python
diab.dropna(inplace=True)
```

```python
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
x = diab.drop('CLASS', axis=1)
y = diab['CLASS']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
smote = SMOTE()
x_train, y_train = smote.fit_resample(x_train, y_train)
```

```python
x_tr , x_val , y_tr , y_val = train_test_split(x_train, y_train, test_size=0.2, random_state=42)
```

```python
std = StandardScaler()
x_tr_std = std.fit_transform(x_tr)
x_val_std = std.transform(x_val)
x_test_std = std.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import roc_auc_score
from sklearn.metrics import roc_curve
from sklearn.metrics import auc , mean_absolute_error , mean_squared_error
from sklearn.metrics import r2_score
```

```python
Log = LogisticRegression()
Log.fit(x_tr_std, y_tr)
y_pred = Log.predict(x_val_std)
print(accuracy_score(y_val, y_pred))
print(mean_absolute_error(y_val, y_pred))
print(mean_squared_error(y_val, y_pred))
print(r2_score(y_val, y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier , DecisionTreeRegressor
```

```python
DTC = DecisionTreeClassifier()
DTC.fit(x_tr_std, y_tr)
y_pred = DTC.predict(x_val_std)
print(classification_report(y_val, y_pred))
print(confusion_matrix(y_val, y_pred))
print(accuracy_score(y_val, y_pred))
```

```python
rfr = DecisionTreeRegressor()
rfr.fit(x_tr_std, y_tr)
y_pred = rfr.predict(x_val_std)
print(mean_absolute_error(y_val, y_pred))
print(mean_squared_error(y_val, y_pred))
print(r2_score(y_val, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier , RandomForestRegressor
```

```python
RFC = RandomForestClassifier()
RFC.fit(x_tr_std, y_tr)
y_pred = RFC.predict(x_val_std)
print(classification_report(y_val, y_pred))
print(confusion_matrix(y_val, y_pred))
print(accuracy_score(y_val, y_pred))
```

```python
RFR = RandomForestRegressor()
RFR.fit(x_tr_std, y_tr)
y_pred = RFR.predict(x_val_std)
print(mean_absolute_error(y_val, y_pred))
print(mean_squared_error(y_val, y_pred))
print(r2_score(y_val, y_pred))
```

```python
from xgboost import XGBClassifier , XGBRegressor
```

```python
XGB = XGBClassifier()
XGB.fit(x_tr_std, y_tr)
y_pred = XGB.predict(x_val_std)
print(classification_report(y_val, y_pred))
print(confusion_matrix(y_val, y_pred))
print(accuracy_score(y_val, y_pred))
```

```python
XGBR = XGBRegressor()
XGBR.fit(x_tr_std, y_tr)
y_pred = XGBR.predict(x_val_std)
print(mean_absolute_error(y_val, y_pred))
print(mean_squared_error(y_val, y_pred))
print(r2_score(y_val, y_pred))
```

```python
XGB_Grid = GridSearchCV(XGBClassifier(), param_grid={'max_depth': [2, 3, 4, 5], 'n_estimators': [50, 100, 150, 200]}, cv=5)
XGB_Grid.fit(x_tr_std, y_tr)
```

```python
import streamlit as st
import pandas as pd
import joblib

st.title('Diabetes Prediction')

# Load your model and dataset
diab = pd.read_csv('diab.csv')
model = joblib.load('XGB_Grid.pkl')

# User chooses text option
gender_text = st.selectbox('Gender', ['F', 'M'])

# Convert text to 0 or 1 as in your dataset
gender_num = 0 if gender_text == 'F' else 1

AGE = st.number_input('AGE')
Urea = st.number_input('Urea')
Cr = st.number_input('Cr')
HbA1c = st.number_input('HbA1c')
Chol = st.number_input('Chol')
TG = st.number_input('TG')
HDL = st.number_input('HDL')
LDL = st.number_input('LDL')
VLDL = st.number_input('VLDL')
BMI = st.number_input('BMI')

if st.button('Predict'):
    # Pass the numeric gender to model
    prediction = model.predict([[gender_num, AGE, Urea, Cr, HbA1c, Chol, TG, HDL, LDL, VLDL, BMI]])
    if prediction == 0:
        st.write('The person is not diabetic')
    else:
        st.write('The person is diabetic')
```
